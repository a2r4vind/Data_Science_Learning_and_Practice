```python
import pandas as pd
import requests
import time
```

```python
# initialize empty list to collect DataFrames
df_list = []

# Use Session object to reuse the same TCP connection (faster and more stable)
with requests.Session() as session:
    
    headers = {
    "accept": "application/json",
    "Authorization": "Bearer eyJhbGciOiJIUzI1NiJ9.eyJhdWQiOiIzYmZjMjhjZGZiYmFhNzBkZjM4NmZkMDk5Y2Q0YjZiMSIsIm5iZiI6MTc5MTc4ODUzOC4yNCwic3ViIjoiNmFjYzg1ZmFjMGZhYTRmODBiYjdiY2JlIiwic2NvcGVzIjpbImFwaV9yZWFkIl0sInZlcnNpb24iOjF9.uzPNtywkob8OS6LXuFA5-ac4FtgrqQVqyWwLgPPnrdU"
    }
    
    session.headers.update(headers)

    for i in range(1,501):
        # update the page in the url with the 'i'
        url = f"https://api.themoviedb.org/3/movie/top_rated?language=en-US&page={i}"

        try:
            response = session.get(url=url)

            # check if request was successful
            if response.status_code == 200:
                data = response.json().get('results',[])
                if data:
                    temp_df = pd.DataFrame(data)[['id','title', 'release_date', 'overview', 'popularity', 'vote_average', 'vote_count']]
                    df_list.append(temp_df)
            else:
                print(f"Failed to fetch page {i}. Status code: {response.status_code}")
        except (requests.exceptions.RequestException, ConnectionError) as e:
            print(f"Connection error on page {i}: {e}. Retrying in 5 seconds...")
            time.sleep(5)
            continue

        time.sleep(0.3)

if df_list:
    df_new = pd.concat(df_list, ignore_index=True)
else:
    df_new = pd.DataFrame()
                
```
